# Task 2 Largest of Three Numbers

```python
a=int(input('Enter Number: '))
b=int(input('Enter Number: '))
c=int(input('Enter Number: '))

if a>b and a>c:
    print(f'{a} is the largest')
elif b>a and b>c:
    print(f'{b} is the largest')
else:
    print(f'{c} is the largest')
```

# Task 3 Sum of Digits

```python
num=123456789
total=0
while num > 0:
    rem=num%10
    total+=rem
    num=num//10
print(total)
```

# Task 4 Reverse a Number

```python
num=123456789
reverse=0
while num > 0:
    rem=num%10
    reverse=reverse*10+rem
    num=num//10
print(reverse)
```

```python
# Part 2: 5 Coding Tasks

# - Task 1 — FizzBuzz: Print numbers from 1 to 50. For multiples of 3 print Fizz, for multiples of 5 print Buzz, and for multiples of both print FizzBuzz.

for i in range(1,51):
    if i % 3 and i % 5 == 0:
        print('Fizz-Buzz')
    elif i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print('Buzz')
    else:
        print(i)
```

```python
dig = int(input("Enter Number: "))

dig = abs(dig)
count = 0

if dig == 0:
    count = 1
else:
    while dig > 0:
        dig = dig // 10
        count += 1

print(count)
```

```python
# Task 3 — Multiplication Table: Take a number from the user and print its multiplication table from 1 to 10. Example: 5 × 1 = 5 through 5 × 10 = 50.

table=int(input('Enter Table: '))
for i in range(1,11):
    print(f'{table}*{i}={table*i}')
```

```python
# Task 4 — Remove Duplicates: Given a list, remove duplicate values without using set(). Example: [1, 2, 2, 3, 4, 4, 5] → [1, 2, 3, 4, 5].

unique=[]
list=[1,2,3,3,4,5,6,6,7,7]
for i in list:
    if i not in unique:
        unique.append(i)
print(unique)
```

```python
# Task 5 — Armstrong Number: Check whether a number is an Armstrong number. Example: 153 → Armstrong Number because 1
# 3
# +5
# 3
# +3
# 3
# # =153. Try solving it without converting the number into a string.


num=1
original=num
total=0

while num > 0:
    rem=num%10
    total+=rem**3
    num=num//10
    
if original==total:
    print('Yes')
else:
    print('Not')
```
