# Task 1 Even or Odd

```python
num=int(input('Enter Number: '))
if num % 2 == 0:
    print('Even')
else:
    print('Odd')
```

# Task 2 Largest of Three Numbers

```python
a=int(input('Enter Number: '))
b=int(input('Enter Number: '))
c=int(input('Enter Number: '))

if a>b and a>c:
    print(f'{a} is the largest')
elif b>a and b>c:
    print(f'{b} is the largest')
else:
    print(f'{c} is the largest')
```

# Task 3 Sum of Digits

```python
num=123456789
total=0
while num > 0:
    rem=num%10
    total+=rem
    num=num//10
print(total)
```

# Task 4 Reverse a Number

```python
num=123456789
reverse=0
while num > 0:
    rem=num%10
    reverse=reverse*10+rem
    num=num//10
print(reverse)
```

# Task 5 Guess the number is +ive -ive or zero

```python
# Guess the number is +ive -ive or zero
a=int(input('Enter Number: '))
if a>0:
    print('Positive')
elif a< 0:
    print('Negative')
else:
    print('Zero')
```
